# The prediction workflow

**Goal.** Set up the place where APEX keeps its services and runs, and connect the Python client to it. A prediction run starts from this setup.

**Prerequisites.**
- APEX installed, with commands typed from the APEX folder as `uv run apex ...`.
- Python packages `numpy`, `pandas` and `matplotlib`, for reading results later.

This notebook does not run a prediction yet, because no example configuration is catalogued for this workflow. It covers the setup every run depends on, so you can add an example once one exists.

## Step 1: Create a services root for this notebook

APEX reads its services from a *services root*, a folder holding files such as `service.toml`. The default location is `~/.apex/service` or `$APEX_HOME`. This notebook uses its own folder, `apex_home`, next to the notebook, so nothing you do here touches your main setup.

The cell below creates the folder.

```python
from pathlib import Path

root = Path("apex_home")
root.mkdir(exist_ok=True)
print("services root:", root, "| exists:", root.is_dir())
```

## Step 2: Connect the Python client

`apex.client.Client` takes the services root and gives you programmatic access to APEX from Python. Create it on the folder from Step 1.

```python
from apex.client import Client

client = Client(Path("apex_home"))
print("client created for", Path("apex_home"))
```

## Reading the results

The cells printed the services root (relative to the notebook) and confirmed that the client was created. If the second cell raised an error, check that the first cell ran and that `apex_home` exists beside the notebook.

## Pitfalls

- **Wrong services root.** A client on `apex_home` sees only what is in that folder. Runs and settings in `~/.apex/service` or `$APEX_HOME` will not appear.
- **Reusing a run name.** A run directory that already exists cannot be started again with `apex run`. Resume it with `apex resume`, or run a copy of the configuration under another `[run]` name.
- **Absolute paths.** Keep notebook paths relative to the notebook folder, so the notebook runs on any machine.
